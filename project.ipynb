```python
import pandas as pd
import re
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
df = pd.read_csv('Keywords_dataset.csv', on_bad_lines='skip') # create a DataFrame 

df.info()
```

```python
def remove_noise(text):
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#\w+', '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'http\S+|www\S+', '', text)         # remove links
    return text;
```

```python
def lower_case(text):
    return text.lower()
```

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
def tokenize_text(text):
    return word_tokenize(text);
```

```python
def remove_stopwords(tokens):
    filters=[];
    for w in tokens:
        if w not in stop_words:
          filters.append(w);
    return filters;
```

```python
def apply_stemm(tokens):
    stemms=[]
    for w in tokens:
        word_stemmer=stemmer.stem(w)
        stemms.append(word_stemmer)
    return stemms;
```

```python
def apply_lemmite(tokens):
    lemmtes=[]
    for w in tokens:
        word_lemmite=lemmatizer.lemmatize(w)
        lemmtes.append(word_lemmite)
    return lemmtes;
```

```python
def preprocess_text(text):
    if not isinstance(text, str):
        return " "
    
    text=remove_noise(text)
    text=lower_case(text)
    text=remove_punctuation(text)
    tokens=tokenize_text(text)
    tokens=remove_stopwords(tokens)
    tokens=apply_stemm(tokens)
    tokens=apply_lemmite(tokens)

    return " ".join(tokens)
```

```python
df['documents'] = df['documents'].astype(str)
df = df.dropna(subset=['documents'])

df['cleaned_dataset_project'] = df['documents'].apply(preprocess_text)
```

```python
import nltk
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import CountVectorizer  # For converting text to numerical features (BoW)

import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

def apply_BoW(texts):

    vectorizer = CountVectorizer(stop_words='english')
    
    vectorizer.fit(texts)
    
    x = vectorizer.transform(texts)
    
    columns = vectorizer.get_feature_names_out()
    
    table = pd.DataFrame(x.todense(), columns=columns, index=texts)
    
    print("Displaying first 3 rows of the Bag of Words table:\n")
    print(table.head(3))  
    
    return table
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

import joblib
```

```python
X_text = df['documents'].astype(str)   # النصوص
y = df['label_names']                  # التصنيف
```

```python
X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text, y, test_size=0.2, random_state=42
)
```

```python
vectorizer = TfidfVectorizer(
    stop_words='english',
    max_features=5000
)

X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)
```

```python
joblib.dump(model, "logistic_text_model.joblib")
joblib.dump(vectorizer, "tfidf_vectorizer.joblib")

print("Model and Vectorizer saved successfully!")
```
